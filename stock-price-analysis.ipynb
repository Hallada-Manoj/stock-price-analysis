```python
# Extract AAPL closing prices
aapl_close = data.loc[data["Name"] == "AAPL", "close"]

# Split data into training and testing sets
split = int(len(aapl_close) * 0.8)

train = aapl_close.iloc[:split]
test = aapl_close.iloc[split:]

print("Total observations:", len(aapl_close))
print("Training observations:", len(train))
print("Testing observations:", len(test))
```

```python
from statsmodels.tsa.arima.model import ARIMA
```

```python
model = ARIMA(train, order=(5, 1, 0))

model_fit = model.fit()

print(model_fit.summary())
```

```python
# Set Date as the time-series index
aapl_close = data.loc[
    data["Name"] == "AAPL"
].set_index("date")["close"]

# Chronological 80:20 split
split = int(len(aapl_close) * 0.8)

train = aapl_close.iloc[:split]
test = aapl_close.iloc[split:]

# Fit ARIMA model again
model = ARIMA(train, order=(5, 1, 0))
model_fit = model.fit()

print(model_fit.summary())
```

```python
predictions = model_fit.forecast(steps=len(test))

predictions = pd.Series(
    predictions.to_numpy(),
    index=test.index
)

print(predictions.head())
```

```python
comparison = pd.DataFrame({
    "Actual Price": test,
    "Predicted Price": predictions
})

comparison.head(10)
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_percentage_error

mae = mean_absolute_error(test, predictions)

mse = mean_squared_error(test, predictions)

rmse = np.sqrt(mse)

mape = mean_absolute_percentage_error(test, predictions) * 100

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("MAPE:", mape, "%")
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    test.index,
    test,
    label="Actual Closing Price"
)

plt.plot(
    predictions.index,
    predictions,
    label="Predicted Closing Price"
)

plt.title("AAPL Actual vs Predicted Stock Prices")
plt.xlabel("Date")
plt.ylabel("Closing Price (USD)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
from statsmodels.tsa.arima.model import ARIMA

# Fit ARIMA using a numerical index
train_reset = train.reset_index(drop=True)

rolling_model = ARIMA(
    train_reset,
    order=(5, 1, 0)
).fit()

rolling_predictions = []

# Predict one step at a time
for i in range(len(test)):

    forecast = rolling_model.forecast(steps=1)

    rolling_predictions.append(forecast.iloc[0])

    # Add actual observation for the next prediction
    rolling_model = rolling_model.append(
        [test.iloc[i]],
        refit=False
    )

# Attach original testing dates
rolling_predictions = pd.Series(
    rolling_predictions,
    index=test.index
)

print(rolling_predictions.head())
```

```python
plt.figure(figsize=(12, 6))

plt.plot(test.index, test, label="Actual Closing Price")

plt.plot(
    rolling_predictions.index,
    rolling_predictions,
    label="One-Step-Ahead Predicted Price"
)

plt.title("AAPL Actual vs One-Step-Ahead Predicted Prices")
plt.xlabel("Date")
plt.ylabel("Closing Price (USD)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
mae_new = mean_absolute_error(test, rolling_predictions)

mse_new = mean_squared_error(test, rolling_predictions)

rmse_new = np.sqrt(mse_new)

mape_new = mean_absolute_percentage_error(
    test, rolling_predictions
) * 100

print("MAE:", mae_new)
print("MSE:", mse_new)
print("RMSE:", rmse_new)
print("MAPE:", mape_new, "%")
```
